## Topic 2 - Four Parts of an AI Agent

1. Brain:
   The LLM that understands messages and makes decisions.

2. Instructions:
   The system prompt that defines business rules,
   responsibilities and response format.

3. Tools:
   Functions that the agent can select and execute
   through the application.

4. Memory:
   Information retained across messages and steps.

### CampusBites Implementation

## Topic 3 - The Agent Loop (ReAct)

ReAct stands for Reasoning and Acting.

An agent loop consists of:
1. Think - Decide the next action.
2. Act - Execute the selected tool.
3. Observe - Receive the tool result.
4. Repeat - Decide whether another action is needed.
5. Stop - Return the final answer or reach the round limit.

The LLM selects the tool, while the application
validates arguments and executes the function.

CampusBites Example:
Customer asks the cost of 3 dosas and 2 teas.

Tool call 1: get_price("dosa", 3)
Observation: 105

Tool call 2: get_price("tea", 2)
Observation: 20

Final answer: Rs 125.

A maximum round limit prevents unnecessary
execution, excessive costs and infinite loops.

## Topic 4 - Three Levels of AI

### Traditional Software
Executes predefined rules written by a programmer.
Example: Calculating the total price of an order.

### Generative AI
Generates content based on user instructions.
Example: Writing CampusBites advertisements.

### Agentic AI
Selects and executes tools to achieve a goal.
Example: Checking food availability, calculating
prices and processing an authorised customer order.

A real-world application can combine all three.

Traditional = Follows rules
Generative = Creates content
Agentic = Decides and acts

## Topic 5 - Agent or Not?

Always choose the simplest solution that satisfies
the requirements.

Plain Code:
Use when the execution steps are predefined.

One LLM Call:
Use when the task requires understanding or
generating content in a single request.

AI Agent:
Use when the system must dynamically select
actions and tools to achieve a goal.

```python
SYSTEM = """
ROLE:
You are the friendly customer support assistant
for CampusBites.

FACTS:
Idli costs Rs 20.
Dosa costs Rs 35.
Upma costs Rs 25.
Poori costs Rs 40.
Veg Fried Rice costs Rs 60.
Tea costs Rs 10.
Coffee costs Rs 15.

Opening hours: 8 AM to 5 PM.
Only vegetarian food is available.
Students can pre-order meals for campus pickup.
Bulk orders require two hours' notice.

RULES:
Never invent menu items, prices or offers.
Never promise delivery or exact pickup times.
If information is unavailable, ask the customer
to check with the canteen manager.
Calculate totals carefully.
Never claim an order has been placed or confirmed.

FORMAT:
Reply politely in under 50 words.
Use English or Telugu matching the customer.
"""

print("System prompt created successfully!")
```

```python
from langchain_core.prompts import (
    ChatPromptTemplate,
    MessagesPlaceholder
)

prompt = ChatPromptTemplate.from_messages([
    ("system", SYSTEM),
    MessagesPlaceholder("history"),
    ("human", "{question}")
])

print("Prompt template created successfully!")
```

```python
from langchain_core.output_parsers import StrOutputParser

chain = prompt | llm | StrOutputParser()

history = []

print("CampusBites chain created successfully!")
```

```python
def ask(question: str) -> str:
    """Generate an answer using conversation history."""

    answer = chain.invoke({
        "history": history,
        "question": question
    })

    history.extend([
        ("human", question),
        ("ai", answer)
    ])

    return answer
```

```python
print(ask("Hi, my name is Ravi."))

print(ask("I want three dosas."))

print(ask("Add two cups of tea."))

print(ask("What is my name and total bill?"))
```

```python
for role, message in history:
    print(role, ":", message)

print("Total messages:", len(history))
```

```python
import csv

with open("tests.csv", "r") as file:
    rows = list(csv.DictReader(file))

print(rows[0])

print("Total test cases:", len(rows))
```

```python
def run_tests(test_chain, path: str = "tests.csv"):
    """Evaluate the chatbot using customer test cases."""

    with open(path, "r") as file:
        rows = list(csv.DictReader(file))

    passed = 0
    results = []

    for row in rows:

        question = row["question"]
        expected = row["must_include"]

        answer = test_chain.invoke({
            "history": [],
            "question": question
        })

        ok = expected.lower() in answer.lower()

        if ok:
            passed += 1

        status = "PASS" if ok else "FAIL"

        print(status, "|", question)
        print("Expected:", expected)
        print("Answer:", answer)
        print("-" * 60)

        results.append({
            "question": question,
            "answer": answer,
            "passed": ok
        })

    total = len(rows)
    pass_rate = (passed / total) * 100

    print(f"\n{passed}/{total} passed")
    print(f"Pass rate: {pass_rate:.2f}%")

    return results
```

```python
baseline_results = run_tests(chain)
```

```python
SYSTEM_V2 = SYSTEM.replace(
    "If information is unavailable, ask the customer\n"
    "to check with the canteen manager.",

    'If information is unavailable, explicitly say: '
    '"Please check with the canteen manager." '
    'Do not guess or make unsupported promises.'
)

print(SYSTEM_V2)
```

```python
prompt_v2 = ChatPromptTemplate.from_messages([
    ("system", SYSTEM_V2),
    MessagesPlaceholder("history"),
    ("human", "{question}")
])

chain_v2 = prompt_v2 | llm | StrOutputParser()
```

```python
improved_results = run_tests(chain_v2)
```

```python
def score(results):
    return sum(row["passed"] for row in results)

print("Original prompt:", score(baseline_results), "/ 10")

print("Improved prompt:", score(improved_results), "/ 10")

print("Change:",
      score(improved_results) - score(baseline_results))
```

```python
llm_high = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=1.0
)

chain_high = prompt_v2 | llm_high | StrOutputParser()
```

```python
high_run1 = run_tests(chain_high)
```

```python
high_run2 = run_tests(chain_high)
```

```python
print("Baseline (0.3):",
      score(baseline_results), "/ 10")

print("Improved prompt (0.3):",
      score(improved_results), "/ 10")

print("High temperature run 1 (1.0):",
      score(high_run1), "/ 10")

print("High temperature run 2 (1.0):",
      score(high_run2), "/ 10")
```

```python
MENU = {
    "idli": 20,
    "dosa": 35,
    "upma": 25,
    "poori": 40,
    "veg fried rice": 60,
    "tea": 10,
    "coffee": 15
}
```

```python
def get_price(item: str, qty: int = 1) -> int:
    """Calculate the total price of a menu item in rupees."""

    item = item.lower().strip()

    if item not in MENU:
        raise ValueError("Item is not on the menu.")

    if type(qty) is not int or qty <= 0:
        raise ValueError("Quantity must be a positive integer.")

    return MENU[item] * qty
```

```python
def check_availability(item: str) -> bool:
    """Check whether an item is listed on the CampusBites menu."""

    item = item.lower().strip()

    return item in MENU
```

```python
print(get_price("dosa", 3))
print(get_price("idli", 5))
print(get_price("tea", 2))

print(check_availability("dosa"))
print(check_availability("shawarma"))
```

```python
assert get_price("dosa", 3) == 105
assert get_price("idli", 5) == 100
assert get_price("tea", 2) == 20
assert get_price(" DOSA ", 2) == 70

assert check_availability("dosa") == True
assert check_availability("shawarma") == False

print("All 6 function tests passed!")
```
