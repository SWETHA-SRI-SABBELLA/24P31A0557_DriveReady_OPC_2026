## Practical 3 - Building CampusBites Assistant

```python
SYSTEM = """
ROLE:
You are the customer support assistant for CampusBites.

FACTS:
1. Menu: Idli Rs 20, Dosa Rs 35, Upma Rs 25,
   Poori Rs 40, Veg Fried Rice Rs 60,
   Tea Rs 10 and Coffee Rs 15.
2. We are open from 8 AM to 5 PM.
3. We serve only vegetarian food.
4. Students can pre-order meals for campus pickup.
5. Bulk orders require two hours' notice.

RULES:
1. Never invent menu items, prices or offers.
2. Never promise an exact pickup time.
3. If information is unavailable, say
   "I'll check with the canteen manager."

FORMAT:
Reply politely in under 50 words.
Use English or Telugu based on the customer's language.
"""
```

```python
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    temperature=0.2,
    messages=[
        {"role": "system", "content": SYSTEM},
        {"role": "user", "content": "Do you offer a 50% discount?"}
    ]
)

print(response.choices[0].message.content)
```

## Practical 4 - Temperature Experimentation

### Day 1 Assignment 1

Objective: To observe how temperature affects
the randomness and creativity of LLM responses.

```python
for i in range(3):
    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        temperature=0.2,
        messages=[
            {"role": "system", "content": SYSTEM},
            {"role": "user", "content": "Write one tagline for CampusBites."}
        ]
    )

    print("Attempt", i + 1)
    print(response.choices[0].message.content)
    print()
```

```python
for i in range(3):
    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        temperature=1.2,
        messages=[
            {"role": "system", "content": SYSTEM},
            {"role": "user", "content": "Write one tagline for CampusBites."}
        ]
    )

    print("Attempt", i + 1)
    print(response.choices[0].message.content)
    print()
```

### Observations

- At temperature 0.2, the model generated relatively
  similar taglines with repeated vocabulary.
- At temperature 1.2, the model generated more varied
  and creative taglines.
- Lower temperature does not guarantee accuracy.
- Low temperature is preferable for customer support.
- Higher temperature can be useful for creative marketing.

### Conclusion

Temperature controls the randomness of LLM responses,
not their factual accuracy.

## Practical 7 - English vs Telugu Token Experiment

### Day 1 Assignment 4

Objective: To compare the number of tokens required
to process the same question in different languages.

```python
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    temperature=0.2,
    messages=[
        {"role": "user", "content": "What is the price of a dosa?"}
    ]
)

print("Response:", response.choices[0].message.content)
print("Prompt tokens:", response.usage.prompt_tokens)
print("Completion tokens:", response.usage.completion_tokens)
print("Total tokens:", response.usage.total_tokens)
```

```python
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    temperature=0.2,
    messages=[
        {"role": "user", "content": "ఒక దోస ధర ఎంత?"}
    ]
)

print("Response:", response.choices[0].message.content)
print("Prompt tokens:", response.usage.prompt_tokens)
print("Completion tokens:", response.usage.completion_tokens)
print("Total tokens:", response.usage.total_tokens)
```

### Observations

| Token Type | English | Telugu |
|---|---|---|
| Prompt Tokens | 79 | 80 |
| Completion Tokens | 98 | 151 |
| Total Tokens | 177 | 231 |

The Telugu question consumed one additional prompt token
and 53 additional completion tokens compared to English.

Telugu consumed approximately 30.5% more total tokens
in this experiment.

### Conclusion

Token consumption varies depending on the language,
tokenizer, input text and generated response.
The number of words is not equal to the number of tokens.
